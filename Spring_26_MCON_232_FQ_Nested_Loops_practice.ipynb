## Nested Lists (2D Lists) — Quick Scaffolding

A **nested list** (also called a **2D list**) is a list where each item is itself a list.
We often use nested lists to represent **grids** like seating charts, game boards, or tables.

### 1) A small example (2 rows × 3 columns)

```python
grid = [
    ["EMPTY", "EMPTY", "EMPTY"],
    ["EMPTY", "EMPTY", "EMPTY"]
]
```
### 2) Building a grid with loops (recommended)

To build a grid, we usually make rows one at a time and add them to the chart.
```
def make_chart(rows, cols):
    chart = []
    for r in range(rows):              # outer loop = rows
        row = []
        for c in range(cols):          # inner loop = columns
            row.append("EMPTY")
        chart.append(row)
    return chart
```

### 3) Printing a grid neatly (nested loops)

```
def print_chart(chart):
    for r in range(len(chart)):
        for c in range(len(chart[0])):
            print(chart[r][c], end=" ")
        print()  # new line after each row
```

### 4) Updating one cell
```
chart[2][3] = "Ava"   # row 2, col 3
```

### 5) Common reminders

- The outer loop usually goes through the rows

- The inner loop usually goes through the columns

- Use len(chart) for number of rows

- Use len(chart[0]) for number of columns (assuming all rows are same length)

- Assign at least 6 names to different seats
- Count how many seats are still `"EMPTY"` and print that number

**Optional extension:** Write the seating chart to a file (one row per line).


````python
# F1 code here
def assign_seat(chart, row, col, name):

  chart[row][col] = name

def print_chart(chart):
  print("---------------------")
  print("Seating arrangement")
  print("---------------------")
  for r in range(len(chart)):
    for c in range(len(chart[0])):
      print(f"{chart[r][c]:7}", end = "")
    print()
  print("---------------------")
def main():
  chart = []

  rows = 4
  cols = 5

  for r in range(rows):
    row_list = []
    for c in range(cols):
      row_list.append("EMPTY")
    chart.append(row_list)

  keep_assigning = 'y'
  while keep_assigning == 'y':
    print_chart(chart)
    print("Adding a name:")
    row = int(input("Please enter the row number: "))
    while row < 0 or row > rows:
      row = int(input("Please enter a valid row: "))

    col = int(input("Please enter the column number: "))
    while col < 0 or col > cols:
      col = int(input("Please enter a valid column: "))

    name = input("Please enter the name: ")
    while not name:
      name = input("Please enter a valid name: ")

    assign_seat(chart, row, col, name)

    print_chart(chart)

    keep_assigning = input("Assign another seat? (y = yes): ")

  print_chart(chart)

  empty_seats = 0
  for r in range(rows):
    for c in range(cols):
      if chart[r][c] == "EMPTY":
        empty_seats += 1

  print(f"There are {empty_seats} empty seats.")
main()
````

- Use nested loops to compute:
  - the **overall average temperature**
  - the **highest single reading** in the grid

**Optional extension:** Count how many readings are above a threshold (e.g., above 75).


````python
# F2 answer here
def set_temp(grid, day, time_index, temp):
  grid[day][time_index] = temp

def print_grid(grid):
  print("\nTemperature data:")
  for r in range(len(grid)):
    print(f"Day {r}:")
    for c in range(len(grid[0])):
      print(f"{time_of_day(c)}: {grid[r][c]}")
    print()

def time_of_day(index):
  if index == 0:
    return "morning"
  elif index == 1:
    return "afternoon"
  else:
    return "evening"

def main():
  rows = 7
  cols = 3

  grid = []

  for r in range(rows):
    row_list = []
    for c in range(cols):
      row_list.append(0)
    grid.append(row_list)

  for day in range(rows):
    for time_index in range(cols):
      temp = int(input(f"Please enter the temperature in the {time_of_day(time_index)} of day {day + 1}: "))
      set_temp(grid, day, time_index, temp)
    print()

  print_grid(grid)

  total = 0
  for r in range(rows):
    for c in range(cols):
      total += grid[r][c]

  average = total / (rows * cols)

  highest = grid[0][0]
  for r in range(rows):
    for c in range(cols):
      if grid[r][c] > highest:
        highest = grid[r][c]

  print(f"Average temperature: {average}")
  print(f"Highest temperature: {highest}")
main()
````
